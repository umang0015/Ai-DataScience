```python
# Create a line chart to visualize the monthly sales trend of a company. Add an appropriate title, axis labels, markers, and grid lines.
company_price = df.groupby("Company")["Price"].mean()
company_price.plot(kind="line" , marker="o" , figsize=(10,5))
plt.title("Average Laptop Price by Company")
plt.xlabel("Company")
plt.ylabel("Avg Price")
plt.grid(True)
plt.show()
```

```python
# Create a multiple line chart to compare the monthly sales of two products (Laptop and Mobile). Include a legend, different markers, and grid lines.
sales_data = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
              "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"],
    "Laptop": [120, 150, 180, 170, 210, 250, 240, 260, 280, 300, 320, 350],
    "Mobile": [100, 130, 160, 190, 220, 240, 260, 250, 270, 290, 310, 330]
})

plt.figure(figsize=(10,5))

plt.plot(sales_data["Month"],
         sales_data["Laptop"],
        marker = "o",
         label="Laptop"
         )
plt.plot(sales_data["Month"] ,
         sales_data["Mobile"],
         marker = "s",
         label="Mobile"
         )
plt.title("Comparison between Laptop and Mobile Chart")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
# Create a vertical bar chart to display the number of employees in different departments. Customize the chart with different bar colors, title, and axis labels.
employee_data = pd.DataFrame({
    "Department": ["HR", "Sales", "IT", "Finance", "Marketing"],
    "Employees": [20, 35, 50, 25, 30]
})
print(employee_data)

plt.figure(figsize=(8,5))
plt.bar(employee_data['Department'] ,
        employee_data['Employees'] ,
        color=["red", "blue", "green", "orange", "purple"])

plt.title("Number of Employees in different department")
plt.xlabel("Department")
plt.ylabel("Employees")
plt.show()
```

```python
# Create a horizontal bar chart to compare the profit earned by different cities. Add suitable chart labels and a title.
profit_data = pd.DataFrame({
    "City": ["Delhi", "Mumbai", "Bangalore", "Chennai", "Hyderabad"],
    "Profit": [25000, 32000, 28000, 21000, 30000]
})

print(profit_data)

plt.figure(figsize=(8,5))

plt.barh(profit_data["City"] ,
         profit_data["Profit"],
         color=["red", "blue", "green", "orange", "purple"]
         )

plt.title("Horizontal bar chart : comparison the profit earned ")
plt.xlabel("Profit")
plt.ylabel("City")
plt.show()
```
