```python
# build simple neural network model for below data
import tensorflow as tf
import numpy as np
x = np.array([1,2,3,4,5])
y = np.array([5,6,7,8,9])
print(x, y)
```

```python
# build the NN architecture with keras
from tensorflow import keras
#to define model type
from keras.models import Sequential
#to add input and output neurons to model (definining input and output in activation function)
from keras.layers import Dense


# define the architecture
model= Sequential()
model.add(Dense(units=1 , input_shape=(1,)))
#units =1 means output neuron is 1 , input shape define actual no. of column
model.summary()
#  define optimizer and metrics
model.compile(optimizer='sgd' , loss=['mean_squared_error'])
```

```python
model.fit(x,y , epochs=100)
```

```python
# print the model weights
print(model.get_weights())
```

```python
import numpy as np
# predict on sample data
print(model.predict(np.array([3])))
```

```python
# save the model
model.save('/content/drive/MyDrive/Models/my_model.h5')
```
