```python
# build the simple n gram language model on sample corpus
# which can predict next possible word
```

```python
# define the small corpus
sentences = [
    "i love machine learning" ,
    "i love deep learning" ,
    "machine learning is useful" ,
    "machine learning is interesting" ,
    "deep learning is powerful" ,
    "deep learning is interesting" ,
    "artificial intelligence is useful",
    "artificial intelligence is powerful"
]

for s in sentences:
  print(s)
```

```python
# tokenize the sentences
tokenized_sentences = [[word for word in sentence.split()] for sentence in sentences]
for s in sentences:
  print(s)
```

```python
# Create a bigrams
bigrams = []
for sentence in tokenized_sentences:
  for i in range(len(sentence) - 1):
    bigrams.append((sentence[i] , sentence[i + 1]))

# count the bigrams using counter
from collections import Counter
bigram_counts = Counter(bigrams)
# print the bigrams
for bigram, count in bigram_counts.items():
  print(bigram, count)

# bigrams count is needed for next step to compute probability
```

```python
# calculate conditional probabilities
# for bigram (W1,W2:
# P(W2| W)) = Count(W1, W2) / count(W1)

context_counts = Counter()
for w1 , w2 in bigrams:
  context_counts[w1] += 1
bigram_probabilities = {}
for (w1 ,  w2) ,count in bigram_counts.items():

  bigram_probabilities[(w1,w2)] = count/context_counts[w1]



# next we will compute the pairwise probabilities
for pair , probability in bigram_probabilities.items():
  print(pair , probability)

  # isme pairs banaya jisme jiski probability jyada hai saath me unko liya
```

```python
#predict next possible word:
def predict_next_word(word):
  candidates={}
  for (w1,w2),probability in bigram_probabilities.items():
    if w1 == word:
      candidates[w2] = probability
  if not candidates:
    return None, None
  best_candidate = max(candidates, key=candidates.get)
  return best_candidate, candidates

  # isme yeh next word predict karega
```

```python
word , candidates = predict_next_word("machine")
print(word)
print(candidates)

# machine ke baad learning agaya
```

```python
# try to generate a sequence , short one with 4 to 8 words

def generate_sequence(word,length):
  # length aur words leliye
  result=[word]
  for i in range (length):
    # length tak loop
    word, _=predict_next_word(word)
    if word is None:
      break
      # kuch nahi mila to break
    result.append(word)
    # appending the word into result
    sentence=" ".join(result)

  return sentence


```

```python
print(generate_sequence("intelligence" , 4))
```
