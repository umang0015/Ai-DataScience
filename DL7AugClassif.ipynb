```python
import pandas as pd
df= pd.read_csv('/content/Admission_Predict.xlsx - Admission_Predict.csv')
df.head()
```

```python
x= df.iloc[:,1:-1]
y = df.iloc[:,-1:]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

x_train,x_test,y_train,y_test= train_test_split(x,y,test_size=0.2,random_state=0)
scaler = StandardScaler()
x_train= scaler.fit_transform(x_train)
x_test= scaler.transform(x_test)
```

```python
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense
```

```python
# hidden layers is 2^n if the input is 7 then the hidden layer is 14 but 16 is nearest to 2^n  so now 16 hidden layers
model = Sequential()
model.add(Dense(units=16 , input_shape=(7,) , activation='sigmoid'))
model.add(Dense(units=24 , activation='sigmoid'))
model.add(Dense(units=1 , activation='sigmoid'))


# as it is binary classication use 1 output neuron
# compile
model.compile(optimizer='adam' , metrics=['accuracy'] ,loss='binary_crossentropy')
model.summary()
```

```python
model.fit(x_train,y_train,epochs=10)
```

```python
# confusion matrix
from sklearn.metrics import confusion_matrix
y_pred = model.predict(x_test)
y_pred = y_pred>0.5
cm = confusion_matrix(y_test,y_pred)
cm
```
