##**Build RNN next seq Predictor**

```python
text = """machine learning is powerful and useful
machine learning helps computers learn from data
deep learning is a part of machine learning
deep learning uses neural networks
natural language processing helps computers understand language
machine learning is used in natural language processing
artificial intelligence is changing many industries
data science uses machine learning for prediction"""
```

```python
# import libraries
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
# agar tokens kam ho toh usme pad laga kar add kar dete hai
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense , SimpleRNN
```

```python
# tokenize , create sequences
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])
total_words= len(tokenizer.word_index) + 1
sequences =[]
for line in text.split("\n"):
  tokens = tokenizer.texts_to_sequences([line])[0]
  for i in range(1 , len(tokens)):
    sequences.append(tokens[:i+1])
# print(sequences)
max_len = max(len(s) for  s in sequences)
# print(max_len)
# max len is 8

# pad seq below max_len
sequences = np.array(pad_sequences(sequences, maxlen = max_len , padding = "pre"))
X = sequences[: , :-1]
y = sequences[: , -1]
# pre padding is used if the curr_len =5 then the 0 added to the sentences at beginnig
# so the predictions will be return true answer
print(X.shape)
print(y.shape)

```

```python
# build an rnn model
rnn_model = Sequential([
    Embedding(total_words ,32 , input_length =max_len-1),
    SimpleRNN(64),
    Dense(total_words , activation="softmax")

])

rnn_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy', metrics=['accuracy'] )
rnn_model.fit(X,y,epochs=100)
```

```python
# build the lstm
lstm_model = Sequential([
    Embedding(total_words ,32 , input_length =max_len-1),
    LSTM(64),
    Dense(total_words , activation="softmax")

])

lstm_model.compile(optimizer='adam',loss='sparse_categorical_crossentropy', metrics=['accuracy'] )
lstm_model.fit(X,y,epochs=100 , verbose=0)
# verbose meaning the training is completed but it does not see the progress
```

```python
# do the predicition
def predict_next_word(model , seed):
  tokens = tokenizer.texts_to_sequences([seed])[0]
  tokens=pad_sequences([tokens] , maxlen = max_len-1 , padding="pre")
  prediction =model.predict(tokens)
  predicted_word = tokenizer.index_word[np.argmax(prediction)]
  return predicted_word
```

```python
model1 = lstm_model
model2 = rnn_model
seed = "natural language processing"

print(predict_next_word(model1 , seed))
print(predict_next_word(model2 , seed))

# full text is natural language processing helps computers understand language
# both show the same result
```
