```python
# next apply preprocessing onto the data
# import necessary libraries first
import nltk
import re
import string
import numpy as np
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize
# ml
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score , confusion_matrix , classification_report

```

```python
# download necessary corpus and docs
nltk.download('stopwords')
nltk.download('punkt')
nltk.download('wordnet')
```

```python
# now applying preprocessing
# build preprocessing function
def text_process(text):
  # convert the text into lower
  text = text.lower()
  # remove the numbers using re
  text = re.sub(r'\d+','',text)
  # remove punctuation
  text = text.translate(str.maketrans('','',string.punctuation))
  # remove white spaces
  text = text.strip()

  # now tokenize the text
  tokens = text.split()
  # remove stopwords

  final_tokens = [word for word in tokens if word not in stopwords.words('english')]
  # applying stemming with porter stemmer
  ps = PorterStemmer()
  final_tokens = [ps.stem(word) for word in final_tokens]
  # join the tokens
  text = ' '.join(final_tokens)
  return text
```

```python
# now make call onto the text processing function by passing message
df['cleaned_message']= df['Message'].apply(text_process)
df.head(2)
```

```python
# apply countvectorizer(Bow ) onto the cleaned message
from sklearn.feature_extraction.text import CountVectorizer , TfidfVectorizer
BoW_vect = CountVectorizer()
X1 =BoW_vect.fit_transform(df['cleaned_message'])
Y1 =df['Category']
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vect = TfidfVectorizer()
X2 = tfidf_vect.fit_transform(df['cleaned_message'])
Y2 =df['Category']
```

```python
# split the training set start with X1 , Y1
X_train , X_test , y_train , y_test = train_test_split(X1,Y1,test_size=0.2,random_state=42)
```

```python
# build the classification model using logistic regression
LRC = LogisticRegression()
LRC.fit(X_train , y_train)
```

```python
# test on test dataset
y_pred= LRC.predict(X_test)
```

```python
# compute accuracy score , confusion matrix
accuracy_score(y_test,y_pred)
```

```python
# now applying wordvec inplace of tfidf/BOW
import gensim
from gensim.models import Word2Vec
```

```python
df['tokens'] =df['cleaned_message'].apply(lambda x : x.split())
df['tokens'].head()
```

```python
# tokenize
# pass to word2vec for training
# next pass the embedding to the ml model
# validate the results
```

```python
w2v_model = Word2Vec(df['tokens'], vector_size=100 ,window=5 , min_count=1 , workers=4 )
print('vocab size' , len(w2v_model.wv))
print('vector size' ,w2v_model.wv.vector_size)
```

```python
# word2vec gives one vector per word , we need it for vector for a message(each row)
# define function which converts a vector to a word to vector for a message
def get_doc_vector(tokens , model):
  vectors = [model.wv[word] for word in tokens if word in model.wv]
  if vectors:
    return np.mean(vectors,axis=0)
  else:
    return np.zeros(model.wv.vector_size)
# call the function
X3 = np.array([get_doc_vector(tokens , w2v_model) for tokens in df['tokens']])
Y3 = df['Category']
```

```python
# split with train test split
X_train , X_test , y_train , y_test = train_test_split(X3,Y3,test_size=0.2,random_state=42)
```

```python
# train on logistic regression
LRC = LogisticRegression()
LRC.fit(X_train , y_train)
y_pred= LRC.predict(X_test)
print(accuracy_score(y_test,y_pred))
confusion_matrix(y_test,y_pred)
```

```python
from sklearn.naive_bayes import GaussianNB

# trying with naive bayes (GaussianNB suitable for continuous, potentially negative data)
GB = GaussianNB()
GB.fit(X_train , y_train)
y_pred = GB.predict(X_test)
print(accuracy_score(y_test,y_pred))
confusion_matrix(y_test,y_pred)
```
