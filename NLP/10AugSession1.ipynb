```python
# bag of words implementation using count vectorizer on a sample document
d1= "the cat sat on the mat"
d2= "the dog sat on the log"
d3= "the cat chased the dog"
```

```python
docs = set([d1,d2,d3])
```

```python
# build the document vocabulary
vocab = [w for d in docs for w in d.split()]
vocab = set(vocab)
vocab = sorted(vocab)
print(vocab)
```

```python
# build a bow matrix
matrix = []
for d in docs:
    row=[]
    for w in vocab:
        row.append(d.count(w))
    matrix.append(row)
print(matrix)
```

```python
# now implement bow on docs using built in count vectorizer
from sklearn.feature_extraction.text import CountVectorizer
cv  = CountVectorizer()
x = cv.fit_transform(docs)
print(x.toarray())
print(cv.get_feature_names_out())
```

```python
# TFIDF implementation manual
d1= "the cat sat on the mat"
d2= "the dog sat on the log"
d3= "the cat chased the dog"
docs = [d1,d2,d3]
```

```python
N = len(docs)
tokens = [doc.split() for doc in docs]
print(tokens)
```

```python
# example term t
term ='mat'
```

```python
# calculate the TF for above word
for i ,toks in  enumerate(tokens):
  tf = toks.count(term)/len(toks)
  print('tf: ', tf)
```

```python
# compute the df and idf for the given term
import math
dfreq = sum(1 for toks in tokens if term in toks)
idf = math.log(N/dfreq)+1
print('idf: ' , idf)
```

```python
# compute tf idf
for i , toks in enumerate(tokens):
  tf = toks.count(term)/len(toks)
  tfidf = tf*idf
  print('tfidf: ', tfidf)
```

```python
# using sklearn to find tfidf
# by default tfidf it uses normalization thats why output is different in manual
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf = TfidfVectorizer()
x= tfidf.fit_transform(docs)
print(x.toarray())
print(tfidf.get_feature_names_out())
# display output as dataframe
import pandas as pd
df = pd.DataFrame(x.toarray(), columns=tfidf.get_feature_names_out())
df
```

```python
tfidf = TfidfVectorizer(smooth_idf=False, norm=None)
x= tfidf.fit_transform(docs)
import pandas as pd
df=pd.DataFrame(x.toarray() , columns=tfidf.get_feature_names_out())
df
```
