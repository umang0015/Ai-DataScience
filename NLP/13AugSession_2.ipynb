## **Neural language Model on small corpus**

- Convert the words (tokens) into tokens(integer) ids
- Create context target pairs(input and output)
- Build embedding layer for word representation
- Use  NN to train and predict

```python
# define the small corpus
sentences = [
    "i love machine learning" ,
    "i love deep learning" ,
    "machine learning is useful" ,
    "machine learning is interesting" ,
    "deep learning is powerful" ,
    "deep learning is interesting" ,
    "artificial intelligence is useful",
    "artificial intelligence is powerful"
]

for s in sentences:
  print(s)
```

```python
# build the ids to the tokens
from tensorflow.keras.preprocessing.text import Tokenizer
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)
# as we have tokens now  , convert them to ids

word_index = tokenizer.word_index #returns the ids
vocab_size = len(word_index) +1
print(vocab_size)
print(word_index)
```

###Creating Input and output pairs
###- context and target pairs
###- example "deep learning is powerful"
###- (4 ,1 ) target 2 (deep(4) ,learning(1) -- > is(2)
### - (1,2) target (9)

```python
# create context -Target Pairs
x = []
y = []

for sentence in sentences:
  word = sentence.split()
  for i in range(2,len(word)): # for making pairs of 0,1,2
    context = word[i-2:i] #this will take input value either 1 or 2
    target = word[i]
    x.append([word_index[w] for w in context])
    y.append(word_index[target])
print(x)
print(y)

# next step for training on NN architecture convert list to np.array
```

```python
# convert to numpy array
import numpy as np
x = np.array(x)
y = np.array(y)
print(x.shape)
print(y.shape)
```

 # neural networks doing the embeddings
# and then embeddings will train not the tokens
 pehle word ko embedded kiya then usko flatten karenge taaki input 1d me hojaye
fir usme dense layer lagayenge

```python
from tensorflow.keras.layers import Embedding , Dense , Input , Flatten
from tensorflow.keras.models import Sequential
# neural networks doing the embeddings
# and then embeddings will train not the tokens
model = Sequential(
    #  pehle word ko embedded kiya then usko flatten karenge taaki input 1d me hojaye
# fir usme dense layer lagayenge
[Embedding(input_dim = vocab_size , output_dim = 10 , input_length =2),
Flatten(),
Dense(units=32 , activation = 'softmax')])

# sparse automatically sparse function karlega jab input value jyada hoti hai like 4 or more
# binary me sirf 2 chaiye hote hai

model.compile(optimizer ="adam" , loss="sparse_categorical_crossentropy" , metrics = ['accuracy'])
model.fit(x,y,epochs=100)
```
