```python
# setup
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM , pipeline ,set_seed
# check for gpu if nor available use cpu
device  = "cuda" if torch.cuda.is_available() else "cpu"
print(device)
```

```python
# tokenization on BERT and GPT models
bert_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
gpt2_tokenizer = AutoTokenizer.from_pretrained("gpt2")

sentences = [
    "the unbelievable transformer arhitecture revolutionized the AI , NLP ,Genai"
    "I can't tell tokenization can handle subwords better"
]
# applying tokenization and print tokens
for sentence in sentences:
  bert_tokens = bert_tokenizer.tokenize(sentences)
  gpt_tokens = gpt2_tokenizer.tokenize(sentences)
  print("Bert tokens =" , bert_tokens)
  print("GPT tokens =" , gpt_tokens)
  print()
```

```python
# sampling control: Temperature
# run same prompt with different temperature settings
gen_tokenizer = AutoTokenizer.from_pretrained("gpt2")
gpt_model  = AutoModelForCausalLM.from_pretrained("gpt2")
set_seed(42)
prompt = "The ocean at sunset is "
input = gen_tokenizer(prompt , return_tensors = "pt")
print(input)
# tokenizing the input prompt is completed ,
```

```python
#  next ask same model same question with different temperatures
out = gpt_model.generate(
    **input,
    max_new_tokens = 25 ,
    do_sample=True , temperature = 0.1
)

# print the decoded tokens
print(gen_tokenizer.decode(out[0]))
out = gpt_model.generate(
    **input,
    max_new_tokens = 25 ,
    do_sample=True , temperature = 0.7
)

# print the decoded tokens
print(gen_tokenizer.decode(out[0]))
out = gpt_model.generate(
    **input,
    max_new_tokens = 25 ,
    do_sample=True , temperature = 0.7
)

# print the decoded tokens
print(gen_tokenizer.decode(out[0]))
out = gpt_model.generate(
    **input,
    max_new_tokens = 25 ,
    do_sample=True , temperature = 1.2
)

# print the decoded tokens
print(gen_tokenizer.decode(out[0]))
out = gpt_model.generate(
    **input,
    max_new_tokens = 25 ,
    do_sample=True , temperature = 1.2
)

# print the decoded tokens
print(gen_tokenizer.decode(out[0]))

out = gpt_model.generate(
    **input,
    max_new_tokens = 25 ,
    do_sample=False
)

# print the decoded tokens
print(gen_tokenizer.decode(out[0]))
```

```python
print(gen_tokenizer.decode(out[0]))
out = gpt_model.generate(
    **input,
    max_new_tokens = 25 , top_k =10 ,
    do_sample=True , temperature = 0.7
)

# print the decoded tokens
print(gen_tokenizer.decode(out[0]))
```

```python
# computing the cost of tokens

price_per_1M_input = 3
price_per_1M_output=10
prompt1 = [
    "the unbelievable transformer arhitecture revolutionized the AI , NLP ,Genai"
    "I can't tell tokenization can handle subwords better"
]

prompt2 = ("you are python trainer , Explain the concept of list comprehension ," "and give the reason why that error is coming")

cost_tokenizer = AutoTokenizer.from_pretrained("gpt2")
tokens1=cost_tokenizer(prompt1)
tokens2 =cost_tokenizer(prompt2)
# compute the cost of both input tokens
cost1 = price_per_1M_input * len(tokens1["input_ids"])/1000000
cost2 = price_per_1M_input * len(tokens2["input_ids"])/1000000
print("Cost of prompt 1 =" , cost1)
print("Cost of prompt 2 =" , cost2)

# now compute the output tokens cost by passing prompt 2 as input
inputs = cost_tokenizer(prompt2 , return_tensors="pt")
out = gpt_model.generate(
    **inputs,
    max_new_tokens = 25 ,
    do_sample=True , temperature = 0.7
)
cost_out = price_per_1M_output * len(out[0])/1000000
print("Cost of output =" , cost_out)
```

```python
# hallucinate the model by asking unknown topic
prompt = ("summarize the key findings of Dr Abdul Kalam's 2021 study on corona Vaccination")
input = gen_tokenizer(prompt , return_tensors = "pt")
out = gpt_model.generate(
    **input,
    max_new_tokens = 25 ,
    do_sample=True , temperature = 0.7
)
print(gen_tokenizer.decode(out[0]))
```

## Adapting Examples for Qwen2.5 and Qwen 0.5 Models

First, let's load the tokenizers and models for Qwen2.5 and Qwen 0.5. We'll use `Qwen/Qwen1.5-2.5B-Chat` for Qwen2.5 and `Qwen/Qwen1.5-0.5B-Chat` for Qwen 0.5, as these are common variants.

```python
# Load Qwen 2.5B and Qwen 0.5B tokenizers and models

# For Qwen 2.5
qwen2_5_tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen1.5-2.5B-Chat")
qwen2_5_model = AutoModelForCausalLM.from_pretrained("Qwen/Qwen1.5-2.5B-Chat", torch_dtype="auto", device_map="auto")
qwen2_5_model.to(device)

# For Qwen 0.5
qwen0_5_tokenizer = AutoTokenizer.from_pretrained("Qwen/Qwen1.5-0.5B-Chat")
qwen0_5_model = AutoModelForCausalLM.from_pretrained("Qwen/Qwen1.5-0.5B-Chat", torch_dtype="auto", device_map="auto")
qwen0_5_model.to(device)

print("Qwen models and tokenizers loaded successfully and moved to", device)
```

### Tokenization with Qwen Models

Let's apply the same sentences to the Qwen tokenizers to see how they handle tokenization compared to BERT and GPT-2.

```python
sentences = [
    "the unbelievable transformer arhitecture revolutionized the AI , NLP ,Genai",
    "I can't tell tokenization can handle subwords better"
]

for sentence in sentences:
  qwen2_5_tokens = qwen2_5_tokenizer.tokenize(sentence)
  qwen0_5_tokens = qwen0_5_tokenizer.tokenize(sentence)
  print(f"\nSentence: {sentence}")
  print("Qwen2.5 tokens =", qwen2_5_tokens)
  print("Qwen0.5 tokens =", qwen0_5_tokens)
```

### Sampling Control (Temperature and Top-k) with Qwen Models

Now, let's test the sampling control by generating text with different `temperature` and `top_k` settings for both Qwen models using the same prompt as before.

```python
prompt = "The ocean at sunset is "

print("\n--- Qwen2.5 Model Generations ---")

# Qwen2.5 - Temperature 0.1
input_qwen2_5 = qwen2_5_tokenizer(prompt, return_tensors="pt").to(device)
out_qwen2_5_t01 = qwen2_5_model.generate(
    **input_qwen2_5,
    max_new_tokens=25,
    do_sample=True,
    temperature=0.1
)
print("Qwen2.5 (T=0.1):", qwen2_5_tokenizer.decode(out_qwen2_5_t01[0], skip_special_tokens=True))

# Qwen2.5 - Temperature 0.7
out_qwen2_5_t07 = qwen2_5_model.generate(
    **input_qwen2_5,
    max_new_tokens=25,
    do_sample=True,
    temperature=0.7
)
print("Qwen2.5 (T=0.7):", qwen2_5_tokenizer.decode(out_qwen2_5_t07[0], skip_special_tokens=True))

# Qwen2.5 - Temperature 1.2
out_qwen2_5_t12 = qwen2_5_model.generate(
    **input_qwen2_5,
    max_new_tokens=25,
    do_sample=True,
    temperature=1.2
)
print("Qwen2.5 (T=1.2):", qwen2_5_tokenizer.decode(out_qwen2_5_t12[0], skip_special_tokens=True))

# Qwen2.5 - Top-k 10, Temperature 0.7
out_qwen2_5_tk10 = qwen2_5_model.generate(
    **input_qwen2_5,
    max_new_tokens=25,
    top_k=10,
    do_sample=True,
    temperature=0.7
)
print("Qwen2.5 (Top-k=10, T=0.7):", qwen2_5_tokenizer.decode(out_qwen2_5_tk10[0], skip_special_tokens=True))


print("\n--- Qwen0.5 Model Generations ---")

# Qwen0.5 - Temperature 0.1
input_qwen0_5 = qwen0_5_tokenizer(prompt, return_tensors="pt").to(device)
out_qwen0_5_t01 = qwen0_5_model.generate(
    **input_qwen0_5,
    max_new_tokens=25,
    do_sample=True,
    temperature=0.1
)
print("Qwen0.5 (T=0.1):", qwen0_5_tokenizer.decode(out_qwen0_5_t01[0], skip_special_tokens=True))

# Qwen0.5 - Temperature 0.7
out_qwen0_5_t07 = qwen0_5_model.generate(
    **input_qwen0_5,
    max_new_tokens=25,
    do_sample=True,
    temperature=0.7
)
print("Qwen0.5 (T=0.7):", qwen0_5_tokenizer.decode(out_qwen0_5_t07[0], skip_special_tokens=True))

# Qwen0.5 - Temperature 1.2
out_qwen0_5_t12 = qwen0_5_model.generate(
    **input_qwen0_5,
    max_new_tokens=25,
    do_sample=True,
    temperature=1.2
)
print("Qwen0.5 (T=1.2):", qwen0_5_tokenizer.decode(out_qwen0_5_t12[0], skip_special_tokens=True))

# Qwen0.5 - Top-k 10, Temperature 0.7
out_qwen0_5_tk10 = qwen0_5_model.generate(
    **input_qwen0_5,
    max_new_tokens=25,
    top_k=10,
    do_sample=True,
    temperature=0.7
)
print("Qwen0.5 (Top-k=10, T=0.7):", qwen0_5_tokenizer.decode(out_qwen0_5_tk10[0], skip_special_tokens=True))
```

### Computing Token Costs with Qwen Models

Let's calculate the theoretical cost of tokens for the Qwen models using the same pricing scheme and prompts.

```python
price_per_1M_input = 3
price_per_1M_output = 10

prompt1 = [
    "the unbelievable transformer arhitecture revolutionized the AI , NLP ,Genai",
    "I can't tell tokenization can handle subwords better"
]

prompt2 = ("you are python trainer , Explain the concept of list comprehension ," "and give the reason why that error is coming")

print("\n--- Qwen2.5 Token Costs ---")
tokens1_qwen2_5 = qwen2_5_tokenizer(prompt1)
tokens2_qwen2_5 = qwen2_5_tokenizer(prompt2)

cost1_qwen2_5 = price_per_1M_input * len(tokens1_qwen2_5["input_ids"][0])/1000000 # Assuming prompt1 is a list, taking the first element
cost2_qwen2_5 = price_per_1M_input * len(tokens2_qwen2_5["input_ids"][0])/1000000
print("Cost of prompt 1 (Qwen2.5) =", cost1_qwen2_5)
print("Cost of prompt 2 (Qwen2.5) =", cost2_qwen2_5)

inputs_qwen2_5 = qwen2_5_tokenizer(prompt2, return_tensors="pt").to(device)
out_qwen2_5_cost = qwen2_5_model.generate(
    **inputs_qwen2_5,
    max_new_tokens=25,
    do_sample=True,
    temperature=0.7
)
cost_out_qwen2_5 = price_per_1M_output * len(out_qwen2_5_cost[0])/1000000
print("Cost of output (Qwen2.5) =", cost_out_qwen2_5)


print("\n--- Qwen0.5 Token Costs ---")
tokens1_qwen0_5 = qwen0_5_tokenizer(prompt1)
tokens2_qwen0_5 = qwen0_5_tokenizer(prompt2)

cost1_qwen0_5 = price_per_1M_input * len(tokens1_qwen0_5["input_ids"][0])/1000000 # Assuming prompt1 is a list, taking the first element
cost2_qwen0_5 = price_per_1M_input * len(tokens2_qwen0_5["input_ids"][0])/1000000
print("Cost of prompt 1 (Qwen0.5) =", cost1_qwen0_5)
print("Cost of prompt 2 (Qwen0.5) =", cost2_qwen0_5)

inputs_qwen0_5 = qwen0_5_tokenizer(prompt2, return_tensors="pt").to(device)
out_qwen0_5_cost = qwen0_5_model.generate(
    **inputs_qwen0_5,
    max_new_tokens=25,
    do_sample=True,
    temperature=0.7
)
cost_out_qwen0_5 = price_per_1M_output * len(out_qwen0_5_cost[0])/1000000
print("Cost of output (Qwen0.5) =", cost_out_qwen0_5)
```

### Hallucination Test with Qwen Models

```python
prompt = ("summarize the key findings of Dr Abdul Kalam's 2021 study on corona Vaccination")

print("\n--- Qwen2.5 Hallucination Test ---")
input_qwen2_5_hallucinate = qwen2_5_tokenizer(prompt, return_tensors="pt").to(device)
out_qwen2_5_hallucinate = qwen2_5_model.generate(
    **input_qwen2_5_hallucinate,
    max_new_tokens=25,
    do_sample=True,
    temperature=0.7
)
print("Qwen2.5:", qwen2_5_tokenizer.decode(out_qwen2_5_hallucinate[0], skip_special_tokens=True))

print("\n--- Qwen0.5 Hallucination Test ---")
input_qwen0_5_hallucinate = qwen0_5_tokenizer(prompt, return_tensors="pt").to(device)
out_qwen0_5_hallucinate = qwen0_5_model.generate(
    **input_qwen0_5_hallucinate,
    max_new_tokens=25,
    do_sample=True,
    temperature=0.7
)
print("Qwen0.5:", qwen0_5_tokenizer.decode(out_qwen0_5_hallucinate[0], skip_special_tokens=True))
```
