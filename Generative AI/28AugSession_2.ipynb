```python
from langchain_huggingface import HuggingFacePipeline
from langchain_core.prompts import ChatPromptTemplate

LLM= HuggingFacePipeline(pipeline=generator)
# prompt = "Explain Artificial Intelligence in simple words"
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "you are a helpful AI assistant"),
    ("human" , "Explain{topic} in simple words")])
# chain the prompt with llm
chain = chat_prompt | LLM
response= chain.invoke({"topic" : "AI"})
print(response)
```

## Adding output parser
- StrOutputParser
- it extracts string content from the output of LLM
- It gives clean string that is easier to use in next step of the chain
* Output parser does not necessarily change what the user sees but it makes the LLM Output easier for next component in the LLM chain
-

```python
from langchain_huggingface import HuggingFacePipeline
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

LLM= HuggingFacePipeline(pipeline=generator)
# prompt = "Explain Artificial Intelligence in simple words"
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "you are a helpful AI assistant"),
    ("human" , "Explain{topic} in simple words")])
# now practising on stroutput parser

parser = StrOutputParser()
# chain the prompt with llm -> stroutputParser
chain = chat_prompt | LLM | parser
response= chain.invoke({"topic" : "AI"})
print(response)
```

<!-- RUNNABLES -->
It is the component that can receive an input and  produce an output


```python
# RunnableLambda example
from langchain_huggingface import HuggingFacePipeline
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableLambda

LLM= HuggingFacePipeline(pipeline=generator)
# prompt = "Explain Artificial Intelligence in simple words"
chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "you are a helpful AI assistant"),
    ("human" , "Explain{topic} in simple words")])
# now practising on stroutput parser

# Now define the function which appends string "LLM response is completed"
# to the output of LLM
def append_text(text):
  return text + "\n LLM  response is completed"
# now create runnable lambda
append_runnable = RunnableLambda(append_text)

parser = StrOutputParser()
# chain the prompt with llm -> stroutputParser -> append_runnable
chain = chat_prompt | LLM | parser
response= chain.invoke({"topic" : "AI"})
print(response)
```

```python
# now build chat with memory
# immemorychatmessagehistory , Runnablewithmessagehistory
from langchain_huggingface import HuggingFacePipeline
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.output_parsers import StrOutputParser
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables import RunnableWithMessageHistory

# Assuming LLM and generator are already defined from previous cells (bElMyGT2qm6T)

chat_prompt = ChatPromptTemplate.from_messages([
    ("system", "you are a helpful AI assistant"),
    MessagesPlaceholder(variable_name="history"),
    ("human" , "{question}")
    ])
# now practising on stroutput parser

parser = StrOutputParser()
history = InMemoryChatMessageHistory()
chain = chat_prompt |  LLM | parser
chat_with_memory = RunnableWithMessageHistory(chain ,
                                              lambda session_id: history,
                                              input_messages_key="question" ,
                                              history_messages_key = "history")
config ={"configurable": {"session_id": "Umang-Session" }}
```

```python
response1= chat_with_memory({"question": "My Name is Umang , I am learning AI"},config=config)
print("Q1: " , response1)
```

### Personalized Study Coach with Memory

Now, let's create a specialized `ChatPromptTemplate` for a personalized study coach. This coach will adapt explanations based on a given subject and student level, and crucially, remember the conversation history.

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables import RunnableWithMessageHistory

# Define a new chat prompt for the study coach
chat_prompt_coach = ChatPromptTemplate.from_messages([
    ("system", "You are a personalized study coach. Your goal is to explain concepts clearly, adapting to the student's level ('{level}') and the specific '{subject}'. Remember our conversation and be helpful."),
    MessagesPlaceholder(variable_name="history"),
    ("human" , "{question}")
])

# Create a new history instance for the coach session
history_coach = InMemoryChatMessageHistory()

# Create the chain for the coach
chain_coach = chat_prompt_coach | LLM | parser

# Wrap the chain with RunnableWithMessageHistory for memory
chat_coach_with_memory = RunnableWithMessageHistory(
    chain_coach,
    lambda session_id: history_coach,
    input_messages_key="question",
    history_messages_key="history"
)

# Define a specific session config for the coach
config_coach = {"configurable": {"session_id": "StudyCoach-Session" }}

print("Study coach setup complete!")
```

### First Question to the Study Coach

Let's ask the coach a question about 'Physics' at an 'Intermediate' level.

```python
# Ask the first question to the study coach
coach_response1 = chat_coach_with_memory.invoke(
    {
        "subject": "Physics",
        "level": "Intermediate",
        "question": "Can you explain the concept of quantum entanglement?"
    },
    config=config_coach
)

print("Coach Q1: ", coach_response1)
```

### Follow-up Question to Test Memory

Now, let's ask a follow-up question without explicitly mentioning the subject or level again, to see if the coach remembers the context.

```python
# Ask a follow-up question - the coach should remember the subject and level implicitly
coach_response2 = chat_coach_with_memory.invoke(
    {
        "subject": "Physics", # Still need to pass subject/level even if implicitly known by model
        "level": "Intermediate",
        "question": "How is it different from classical correlation?"
    },
    config=config_coach
)

print("Coach Q2: ", coach_response2)
```
