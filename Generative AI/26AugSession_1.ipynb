```python
# load the llm pipeline with qwen 1.5B model
from transformers import pipeline
import torch
LLM= pipeline("text-generation" , model = "Qwen/Qwen2.5-1.5B-Instruct" , device_map ="auto")
print("Model Loaded")
```

## Zero Shot Prompting
- #### here with give model the instruction alone
## Few shot Prompting
-  #### here we give model with instruction plus an example scenarios


```python
#  this is content
message = """ classify the sentiment as Positive , Negative , or Neutral:\n\"...
 The Delivery was late but the product quality was good"""
# now the role
Prompt= [{"role": "user" , "content":message}]
result = LLM(Prompt , max_new_tokens=100 ,do_sample=True , temperature=0.7)
print(result[0]["generated_text"][-1]["content"])

# the output of the system is in it
# result[0] gives generated result
# and the acutal content stored in [-1]["content"]
# max_new_tokens = 100 means it print the outcomes in 100 tokens
# maximum length = input + output

```

```python
# few shot prompting or few shot learning
message = "classify the sentiment as Positive , Negative , or Neutral:"

Example = """
 "Amazing Service , will buy again." -> Positive\n"
  " terrible packaging, item arrived broken." -> negative\n"
  "'its okay , does the job." -> neutral\n"
  "The delivery was late but the product quality was good." ->"""

Prompt = [{"role": "user" , "content" :message+Example}]
result = LLM(Prompt , max_new_tokens=100 ,do_sample=True , temperature=0.7)
print(result[0]["generated_text"][-1]["content"])
```

```python
# role based prompting
# assigning the model a persona / role that shapes the tone , vocabulary and priorities
Message = " correct the grammar in this sentence:\" Me and him goes to the market yesterday"
Prompt = [{"role":"user", "content":Message}]
result = LLM(Prompt, max_new_tokens=100, do_sample=True, temperature=0.7)
print(result[0]["generated_text"][-1]["content"])
```

```python
# COT , chain of thought
# this helps to understand how the models working step by step on a given task ,mostly to understand , complex ML Architecture ,
# like transformers , the math behind DL
# mathematical problems solving step by step
Message = """ A score had 200 Apples. They sold 30% of them by noon and 20 more by evening , How many apples are left? """
# trying with cot
Prompt= [{"role" : "user" , "content" :Message}]
result = LLM(Prompt , max_new_tokens=200  , do_sample=True , temperature=0.7)
print(result[0]["generated_text"][-1]["content"])
```

```python
Message = """ A score had 200 Apples. They sold 30% of them by noon and 20 more by evening , How many apples are left?
let's think step by step
"""
# trying with cot
Prompt= [{"role" : "user" , "content" :Message}]
result = LLM(Prompt , max_new_tokens=300  , do_sample=True , temperature=0.7)
print(result[0]["generated_text"][-1]["content"])
```

```python
Message = """A former has 17 sheeps. All but 9 die. How many are left?.
 Let's think step by step , then give the final answer as Answer: <number>"""

Prompt = [{"role" : "user" , "content": Message}]
num_samples=5
answers=[]
for i in range(num_samples):
  result = LLM(Prompt , max_new_tokens=300 , do_sample=True , temperature=0.7)
  answers.append(result[0]["generated_text"][-1]["content"])
print(answers)
```

```python
# now collect exact answers and vote for majority
from collections import Counter
extract_ans = []
for ans in answers:
  if "Answer:" in ans:
    final_ans= ans.split("Answer:")[-1].strip()
  extract_ans.append(final_ans)
print(extract_ans)
print(Counter(extract_ans).most_common(1)[0][0])
```
