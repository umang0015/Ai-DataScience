# Hugging Face Pre-Trained Models — 11 Models / 3 Tasks

## Goal

For every model:

1. Load the model.
2. Run three tasks that are **compatible with that model's architecture**.
3. Record output and inference time.
4. Compare models within the same task/model family.

> The 11 models do not all support the same pipelines. Forcing a generation or summarization pipeline onto DistilBERT, for example, would be technically incorrect. This notebook therefore keeps three experiments per model while respecting each model's architecture.

## Imports and Device

````python
from transformers import pipeline
import pandas as pd
import time
import torch

device = 0 if torch.cuda.is_available() else -1

print("GPU available:", torch.cuda.is_available())
if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
````

## Common Inputs

````python
generation_prompt = "Explain artificial intelligence in simple words."

completion_prompt = "The future of artificial intelligence is"

question = "What is artificial intelligence?"

article = '''
Artificial intelligence is a branch of computer science that focuses on
creating systems capable of performing tasks that normally require human
intelligence. These tasks include learning, reasoning, problem solving,
understanding language, recognizing images, and making decisions.
Modern AI systems use machine learning and deep learning techniques to
learn patterns from large amounts of data.
'''

translation_text = "Artificial intelligence is changing the way people work and learn."
````

# MODEL 1 — SmolLM3 3B

**Hugging Face ID:** `HuggingFaceTB/SmolLM3-3B`

**Architecture:** Causal Language Model

### Three compatible tasks
1. Text Generation
2. Text Completion
3. Prompt-Based Question Answering


````python
model_id = "HuggingFaceTB/SmolLM3-3B"

generator = pipeline(
    "text-generation",
    model=model_id,
    device=device
)

print("Loaded:", model_id)
````

## Task 1 — Text Generation

````python
start = time.time()

result = generator(
    generation_prompt,
    max_new_tokens=60,
    num_return_sequences=1,
    do_sample=True
)

print("Time:", round(time.time() - start, 2), "seconds")
print(result[0]["generated_text"])
````

## Task 2 — Text Completion

````python
start = time.time()

result = generator(
    completion_prompt,
    max_new_tokens=50,
    num_return_sequences=1,
    do_sample=True
)

print("Time:", round(time.time() - start, 2), "seconds")
print(result[0]["generated_text"])
````

## Task 3 — Prompt-Based Question Answering

````python
qa_prompt = (
    "Answer this question clearly and briefly. "
    "Question: " + question
)

start = time.time()

result = generator(
    qa_prompt,
    max_new_tokens=60,
    num_return_sequences=1,
    do_sample=False
)

print("Time:", round(time.time() - start, 2), "seconds")
print(result[0]["generated_text"])
````

# MODEL 8 — FLAN-T5 Small

**Hugging Face ID:** `google/flan-t5-small`

**Architecture:** Encoder-Decoder / Text-to-Text

### Three compatible tasks
1. Text-to-Text Generation
2. Summarization
3. Translation


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_id = "google/flan-t5-small"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)

print("Loaded:", model_id)
````

## Task 1 — Text-to-Text Generation

````python
prompt = "Explain artificial intelligence in simple words."

inputs = tokenizer(
    prompt,
    return_tensors="pt"
)

outputs = model.generate(
    **inputs,
    max_new_tokens=60
)

result = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(result)
````

## Task 2 — Summarization

````python
summary_prompt = "summarize: " + article

inputs = tokenizer(
    summary_prompt,
    return_tensors="pt"
)

outputs = model.generate(
    **inputs,
    max_new_tokens=60
)

summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Summary:")
print(summary)
````

## Task 3 — Translation

````python
translation_prompt = (
    "translate English to French: "
    "Artificial intelligence is changing the way people work and learn."
)

inputs = tokenizer(
    translation_prompt,
    return_tensors="pt"
)

outputs = model.generate(
    **inputs,
    max_new_tokens=60
)

translation = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Translation:")
print(translation)
````

# MODEL 10 — BART Large CNN

**Hugging Face ID:** `facebook/bart-large-cnn`

**Architecture:** Encoder-Decoder / Text-to-Text

### Three compatible tasks
1. Text-to-Text Generation
2. Summarization
3. Translation


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
import torch

model_id = "facebook/bart-large-cnn"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForSeq2SeqLM.from_pretrained(model_id)

# Move model to GPU if available
if torch.cuda.is_available():
    model = model.to("cuda")

print("Loaded:", model_id)
````

## Task 1 — Normal summarization

````python
start = time.time()

inputs = tokenizer(
    article,
    return_tensors="pt",
    truncation=True,
    max_length=1024
)

if torch.cuda.is_available():
    inputs = {key: value.to("cuda") for key, value in inputs.items()}

outputs = model.generate(
    **inputs,
    max_new_tokens=60,
    min_new_tokens=20
)

summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Time:", round(time.time() - start, 2), "seconds")
print("Summary:")
print(summary)
````

## Task 2 — short Summarization

````python
start = time.time()

inputs = tokenizer(
    article,
    return_tensors="pt",
    truncation=True,
    max_length=1024
)

if torch.cuda.is_available():
    inputs = {key: value.to("cuda") for key, value in inputs.items()}

outputs = model.generate(
    **inputs,
    max_new_tokens=30,
    min_new_tokens=10
)

short_summary = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Time:", round(time.time() - start, 2), "seconds")
print("Short Summary:")
print(short_summary)
````

## Task 3 — Summarize Another Text

````python
text2 = """
Machine learning is a subset of artificial intelligence that allows
computers to learn patterns from data without being explicitly programmed.
Supervised learning uses labelled data, while unsupervised learning finds
patterns in data without labels. Deep learning is a type of machine
learning based on neural networks with many layers.
"""

start = time.time()

inputs = tokenizer(
    text2,
    return_tensors="pt",
    truncation=True,
    max_length=1024
)

if torch.cuda.is_available():
    inputs = {key: value.to("cuda") for key, value in inputs.items()}

outputs = model.generate(
    **inputs,
    max_new_tokens=50,
    min_new_tokens=15
)

summary2 = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print("Time:", round(time.time() - start, 2), "seconds")
print("Summary:")
print(summary2)
````

# MODEL 11 — DistilBERT SST-2

**Architecture:** Encoder-only classification model

DistilBERT is not a text-generation or summarization model. Therefore, the three experiments below are classification-focused.


````python
model_id = "distilbert/distilbert-base-uncased-finetuned-sst-2-english"

classifier = pipeline(
    "sentiment-analysis",
    model=model_id,
    device=device
)

print("Loaded:", model_id)
````

````python
text = "I really enjoyed this product. The quality is excellent."

start = time.time()
result = classifier(text)

print("Time:", round(time.time() - start, 2), "seconds")
print(result)
````

## Task 2 — Batch Sentiment Classification

````python
texts = [
    "I really enjoyed this product.",
    "The service was terrible.",
    "The movie was excellent and entertaining.",
    "The experience was disappointing."
]

results = classifier(texts)

for text, result in zip(texts, results):
    print("\nText:", text)
    print("Prediction:", result)
````

# Why are the tasks different?

Think of the models as being trained differently.

### 1–7: Causal Language Models

These models predict the **next token**.

For example:

```text
The capital of India is
             ↓
           Delhi
```

So they are naturally good at:

```text
Text Generation
       ↓
Text Completion
       ↓
Question → Generated Answer
```

That's why for these 7 models we have:

### Task 1 — Text Generation

```text
Explain artificial intelligence in simple words.
```

### Task 2 — Text Completion

```text
The future of artificial intelligence is...
```

### Task 3 — Question Answering

```text
What is artificial intelligence?
```

The third one is technically **prompt-based Q&A**, not extractive QA like BERT's QA head. The model generates an answer.

---

# 8–10: T5 / FLAN-T5 / BART

These are **encoder-decoder models**.

They are designed for transforming one piece of text into another.

So we use:

### Task 1 — Text-to-Text

```text
Explain artificial intelligence in simple words.
```

### Task 2 — Summarization

```text
Long article
     ↓
Short summary
```

### Task 3 — Translation

```text
English
   ↓
French
```

For example:

```text
Artificial intelligence is changing the way people work and learn.
```

↓

```text
L'intelligence artificielle change la façon dont les gens travaillent et apprennent.
```

---

# 11: DistilBERT

This one is different.

**DistilBERT is an encoder-only model**, so we shouldn't ask it to generate paragraphs.

It is very good for **classification**.

Therefore we use:

### Task 1 — Sentiment

```text
"I loved this movie."

→ POSITIVE
```

### Task 2 — Batch Sentiment

Give it multiple sentences:

```text
"I loved this movie."
"The service was terrible."
"The product was excellent."
```

and get:

```text
POSITIVE
NEGATIVE
POSITIVE
```

### Task 3 — Confidence Analysis

We examine:

```text
POSITIVE → 0.98
NEGATIVE → 0.95
```
