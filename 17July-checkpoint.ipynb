```python
# ***Outlier Detection***
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = sns.load_dataset('tips')
numeric_cols = ['total_bill' , 'tip' , 'size']
print(f"Original dataset: {df.shape[0]} rows")
```

```python
z_scores = np.abs((tips['tip'] - tips['tip'].mean()) /tips['tip'].std())
print(z_scores)
outliers_zscores = tips[z_scores >3]
print("Outliers detected using Z-score")
print(outliers_zscores)

plt.figure(figsize=(8,5))
plt.scatter(range(len(tips)) , tips['tip'] , label = 'Data' , alpha=0.7)

# highlights outliers 
plt.scatter(outliers_zscores.index , outliers_zscores['tip'] , color='red' , label ='Z-score Outliers')

```

```python
# remove outliers using z -score  
# keep only rows where z-score <3
df_no_outliers_zscores = df[z_scores<3]
print("Data size after removing Z-score outliers: " , len(df_no_outliers_zscores))

```

```python
# Z-scores calculation 
z_scores = np.abs((df['tip'] - df['tip'].mean()) /df['tip'].std())
# remove outliers 
df_no_outliers_zscores = df[z_scores <3]
plt.figure(figsize=(8,5))
plt.scatter(range(len(df_no_outliers_zscores)) ,df_no_outliers_zscores['tip'] , label ='Cleaned Data' ,alpha=0.7 , color='blue')

plt.title('Scatter Plot of Tips(After Z-score Outlier Removal')
plt.xlabel('Index')
plt.ylabel('Tip Amount')
plt.legend()
plt.show()
```

```python
#zcores calculation for total bill
score_bill = np.abs((df['total_bill'] - df['total_bill'].mean()) / df['total_bill'].std())
outliers = df[score_bill > 3]
print("outliers detected using Z-score: ")
print(outliers)
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(df['total_bill'])
plt.title('Boxplot of total bill (Before Outlier Removal)')
plt.ylabel("Total Bill Amount")
plt.show()
```

```python
# calculate Q1 , Q3 and IQR for  'size'
Q1 = df['total_bill'].quantile(0.25)
Q3 = df['total_bill'].quantile(0.75)
IQR = Q3 - Q1

# define bounds
lower_bound = Q1 - 1.5 *IQR 
upper_bound = Q3 + 1.5 *IQR

print(lower_bound)
print(upper_bound)
```

```python
# detect outliers
outliers_iqr = df[(df['total_bill'] < lower_bound) | (df['total_bill'] >upper_bound)]
print("Outliers detected using IQR")
print(outliers_iqr['total_bill'])
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(df['total_bill'])
plt.title("Boxplot of total bill (Before Outlier Removal)")
plt.ylabel("Party size")
plt.show()
```

```python
def remove_iqr_outliers(data,column):
    while True:
        Q1 = data[column].quantile(0.25)
        Q2 = data[column].quantile(0.75)
        IQR = Q3 -Q1 
        lower_bound = Q1 - 1.5 * IQR 
        upper_bound = Q3 + 1.5 * IQR

        # filter data 
        filtered = data[(data[column] >= lower_bound) & (data[column] <=upper_bound)]
        # stop if no change
        if len(filtered) == len(data):
            break
        data  = filtered

    return data

df_no_outliers_iqr_iter = remove_iqr_outliers(df , 'total_bill')
print("Final size after iterative IQR removal: ", len(df_no_outliers_iqr_iter))
```

```python
plt.figure(figsize=(6,4))
plt.boxplot(df_no_outliers_iqr_iter['total_bill'])
plt.title('Boxplot of size (After Iterative IQR Outlier Removal)')
plt.ylabel("Party Size")
plt.show()
```
